```python
import altair as alt

plot1 = alt.Chart(tidy_womensbb_df).mark_bar().encode(
    alt.X("reg_wins_pct:Q", bin = True, title = "Regional Win Percentage"),
    y = alt.Y("count()", title = "Count")
).properties(title = "Distribution of Regional Win Percentage")

plot1
```

```python
plot2 = alt.Chart(tidy_womensbb_df).mark_bar().encode(
    x = alt.X("tourney_finish:N", title = "Tournament Finish", sort = "-y"),
    y = alt.Y("count()", title = "Count"),
    color = "tourney_finish:N"
).properties(title = "Distribution of Tournament Finishes")

plot2
```

Visualization 1: This visualization is relevant to the analysis because it assesses whether teams with higher regional win percentages have a relationship with deeper tournament finishes. It also helps identify if success throughout the regular season correlates with post-season performance.

Visualization 2: This visualization is relevant to the analysis because it shows how conference performance correlates with tournament outcomes. It can help determine if success in a certain conference can be a predictor of team progression in the overall tournament.

```python
viz2 = alt.Chart(tidy_womensbb_df).mark_boxplot().encode(
    x = alt.X("tourney_finish:N", title = "Tournament Finish"),
    y = alt.Y("conf_wins_pct:Q", title = "Conference Win Percentage"),
    color = "tourney_finish:N"
).properties(title = "Conference Win Percentage by Tournament Finish")

viz2
```
